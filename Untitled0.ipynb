```python
# Подключаем библиотеки Pandas, NumPy и Matplotlib
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Загружаем csv-файл в DataFrame с именем df
df = pd.read_csv('/boot/spotify_artist_streaming_2020_2025.csv')

# Проверяем, что данные загрузились
print("Данные успешно загружены. Размер таблицы:", df.shape)
```

```python
# Выводим датасет (первые и последние строки для компактности)
df
```

```python
# Выводим первые 10 строк таблицы
df.head(10)
```

```python
# Выводим названия всех столбцов
print("Названия столбцов:")
print(df.columns.tolist())
```

```python
# Смотрим типы данных столбцов
df.info()
```

```python
# Сортируем по stream_count по убыванию и берём топ-10
top10 = df.sort_values(by='stream_count', ascending=False).head(10)
top10[['track_name', 'artist_name', 'genre', 'stream_count']]
```

```python
# Находим треки, выпущенные после 2024 года
after_2024 = df[df['release_year'] > 2024]
print("Количество треков после 2024 года:", after_2024.shape[0])
after_2024.head()
```

```python
# Среднее и медиана для popularity
mean_pop = df['popularity'].mean()
median_pop = df['popularity'].median()
print("Среднее popularity:", round(mean_pop, 2))
print("Медиана popularity:", median_pop)
```

```python
# Минимум, максимум и стандартное отклонение для popularity
min_pop = df['popularity'].min()
max_pop = df['popularity'].max()
std_pop = df['popularity'].std()
print("Минимум popularity:", min_pop)
print("Максимум popularity:", max_pop)
print("Стандартное отклонение popularity:", round(std_pop, 2))
```

```python
# Мода для genre
mode_genre = df['genre'].mode()
print("Мода для genre:")
print(mode_genre)
```

```python
# Мода для release_year
mode_year = df['release_year'].mode()
print("Мода для release_year:")
print(mode_year)
```

```python
# Строим гистограмму распределения popularity
plt.figure(figsize=(10, 6))
plt.hist(df['popularity'], bins=20, color='steelblue', edgecolor='black')
plt.title('Распределение популярности треков (popularity)')
plt.xlabel('Popularity')
plt.ylabel('Количество треков')
plt.grid(axis='y', alpha=0.3)
plt.show()
```
