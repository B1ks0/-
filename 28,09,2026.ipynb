```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Загрузка данных
df = pd.read_csv("/content/vgsales.csv")

# Первые 10 строк
df.head(10)
```

```python
print("Количество строк:", df.shape[0])
print("Количество столбцов:", df.shape[1])
print("Названия столбцов:", df.columns.tolist())
print("\nТипы данных:")
print(df.dtypes)
print("\nСтатистические характеристики:")
df.describe()
```

```python
# Количество пропусков в каждом столбце
df.isnull().sum()
```

```python
# Строки, где отсутствует Year
df[df["Year"].isnull()]
```

```python
df_clean = df.dropna(subset=["Year"])

print("Размер исходного датасета:", df.shape)
print("Размер очищенного датасета:", df_clean.shape)
print("\nПропуски после очистки:")
print(df_clean.isnull().sum())
```

```python
print("Жанры:", df_clean["Genre"].nunique())
print("Платформы:", df_clean["Platform"].nunique())
print("Издатели:", df_clean["Publisher"].nunique())

print("\nУникальные жанры:")
print(df_clean["Genre"].unique())
```

```python
genre_count = df_clean.groupby("Genre")["Name"].count().sort_values(ascending=False)
print(genre_count)
```

```python
genre_sales = df_clean.groupby("Genre")["Global_Sales"].sum().sort_values(ascending=False)
print(genre_sales)
```

```python
top10 = df_clean.nlargest(10, "Global_Sales")[
    ["Rank", "Name", "Platform", "Genre", "Global_Sales"]
]
print(top10)
```

```python
# Топ-10 платформ по количеству игр
platform_count = df_clean.groupby("Platform")["Name"].count().sort_values(ascending=False)
print("Топ-10 платформ по количеству игр:")
print(platform_count.head(10))

# Топ-10 платформ по суммарным продажам
platform_sales = df_clean.groupby("Platform")["Global_Sales"].sum().sort_values(ascending=False)
print("\nТоп-10 платформ по продажам:")
print(platform_sales.head(10))
```

```python
plt.figure(figsize=(12, 6))
genre_sales.sort_values(ascending=False).plot(kind="bar", color="steelblue")
plt.title("Суммарные мировые продажи по жанрам")
plt.xlabel("Жанр")
plt.ylabel("Продажи, млн копий")
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
corr_na_global = df_clean["NA_Sales"].corr(df_clean["Global_Sales"])
print("Корреляция NA_Sales и Global_Sales:", corr_na_global)

# Или через таблицу
print(df_clean[["NA_Sales", "Global_Sales"]].corr())
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(
    df_clean["NA_Sales"],
    df_clean["Global_Sales"],
    alpha=0.5,
    color="darkorange",
    edgecolors="none"
)
plt.title("Связь продаж в Северной Америке и мировых продаж")
plt.xlabel("Продажи в Северной Америке, млн")
plt.ylabel("Мировые продажи, млн")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
region_corr = df_clean[
    ["NA_Sales", "EU_Sales", "JP_Sales", "Other_Sales", "Global_Sales"]
].corr()

print(region_corr)

# Визуализация
plt.figure(figsize=(8, 6))
plt.imshow(region_corr, cmap="coolwarm", vmin=-1, vmax=1)
plt.colorbar(label="Коэффициент корреляции")
plt.xticks(range(len(region_corr.columns)), region_corr.columns, rotation=45)
plt.yticks(range(len(region_corr.columns)), region_corr.columns)
for i in range(len(region_corr)):
    for j in range(len(region_corr)):
        plt.text(j, i, f"{region_corr.iloc[i, j]:.2f}",
                 ha="center", va="center", color="black")
plt.title("Корреляционная матрица региональных продаж")
plt.tight_layout()
plt.show()
```
